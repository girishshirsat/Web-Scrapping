```python
data=soup.find_all("div",class_="mb-5")
```

```python
len(data)
```

```python
company_name=[]
rating=[]
industry = []
location = []
for i in data:
    #headig
    head=i.find('h2')
    if head is None: #if h2 is not availabe i.e None it will rail error
        pass
    else:
        company_name.append(head.text)

    #rating
    ratng=i.find('div',class_="flex flex-row android:mb-[4px]")
    if ratng is None:
        pass
    else:
        rating.append(ratng.text)

    # Industry and Location
    span = i.find('span', class_="overflow-hidden text-left text-[14px] font-pn-400 leading-5 tracking-[0.25px] text-[#394468] max-md:text-[12px] max-md:leading-4")

    if span is None:
        pass
    else:
        text = span.get_text(strip=True)
        parts = text.split('|', 1)

        if len(parts) == 2:
            industry.append(parts[0].strip())
            location.append(parts[1].split('other locations')[0].strip())
```

```python
df=pd.DataFrame({"name":company_name,
                "rating":rating,
                "industry":industry,
                "location":location})
```

```python
dfs=[]
for j in range(1,10):
    url=f"https://www.ambitionbox.com/list-of-companies?campaign=desktop_nav&page={j}"
    headers={'User-Agent':'Mozilla/5.0 (Windows NT 6.3; Win 64 ; x64) Apple WeKit /537.36(KHTML , like Gecko) Chrome/80.0.3987.162 Safari/537.36'} 
    webpage=requests.get(url,headers=headers)
    print("Page:", j, "Status:", webpage.status_code)
    soup=BeautifulSoup(webpage.text,"html.parser")
    dat = soup.find_all("div", class_="mb-5")
    
    company_name=[]
    rating=[]
    industry = []
    location = []
    for i in dat:
        #headig
        head=i.find('h2')
        if head is None: #if h2 is not availabe i.e None it will rail error
            pass
        else:
            company_name.append(head.text)
    
        #rating
        ratng=i.find('div',class_="flex flex-row android:mb-[4px]")
        if ratng is None:
            pass
        else:
            rating.append(ratng.text)
    
        # Industry and Location
        span = i.find('span', class_="overflow-hidden text-left text-[14px] font-pn-400 leading-5 tracking-[0.25px] text-[#394468] max-md:text-[12px] max-md:leading-4")
    
        if span is None:
            pass
        else:
            text = span.get_text(strip=True)
            parts = text.split('|', 1)
    
            if len(parts) == 2:
                industry.append(parts[0].strip())
                location.append(parts[1].split('other locations')[0].strip())
    df=pd.DataFrame({"name":company_name,
                "rating":rating,
                "industry":industry,
                "location":location})
    dfs.append(df)
if dfs:
    final = pd.concat(dfs, ignore_index=True)
else:
    final = pd.DataFrame()
```
