```python
import pandas as pd
import requests
```

```python
response=requests.get("https://api.themoviedb.org/3/movie/top_rated?api_key=4bc52c529a34c4c2126ce485dd1a6a06&language=en-US&page=1")
```

```python
df=pd.DataFrame(response.json()["results"])[["id","title","overview","release_date","popularity","vote_average","vote_count"]]
```

```python
df
```

```python
dfs=[]
for i in range(1,5):
    re=requests.get("https://api.themoviedb.org/3/movie/top_rated?api_key=4bc52c529a34c4c2126ce485dd1a6a06&language=en-US&page={}".format(i))
    tempdf=pd.DataFrame(re.json()["results"])[["id","title","overview","release_date","popularity","vote_average","vote_count"]]
    dfs.append(tempdf)
df = pd.concat(dfs, ignore_index=True)
```

```python
df
```

```python
df.to_excel("TMDB_TOP_MOVIE.xlsx", index=False)
```
